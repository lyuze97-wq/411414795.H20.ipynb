```python
bag=set()
print("初始背包",bag)
bag.add("刀")
bag.add("盾")
bag.add("藥水")
print("撿到道具後的背包",bag)
bag.add("藥水")
print("再次撿到藥水的背包",bag)
bag.remove("盾")
print("使用盾後的背包",bag)
lose=bag.pop()
print("掉落的道具",lose)
print("遊戲結束,最後背包",bag)
```

```python
def add(x,y):
  try:
    ans=x/y
  except ZeroDivisionError:
    return "除數不可為0"
  else:
    return ans

print(add(10,0))
print(add(10,2))
print(add(6,0))
```
